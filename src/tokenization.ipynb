```python
curr_tok = 256
def minBPE(tokens):
    global curr_tok

    res = []
    occur = {}

    for i in range(len(tokens)-1):
        pair = tokens[i:i+2]
        pair = (pair[0], pair[1])
        occur[pair] = occur.get(pair, 0) + 1
    
    max_occ = max(occur.values())
    max_pair = None
    
    for k, v in occur.items():
        if v == max_occ:
            max_pair = k
            break
    i = 0
    while i < len(tokens) - 1:
        if tokens[i] == max_pair[0] and tokens[i+1] == max_pair[1]:
            res.append(curr_tok)
            i += 2
        else:
            res.append(tokens[i])
            i += 1
    curr_tok += 1
    return res

tokens = minBPE(tokens)
print(len(tokens))
```
